```python
from pyspark.sql import SparkSession

# Paramos qualquer sessão zumbi
try:
    spark.stop()
except:
    pass

# Sessão limpa e sem comentários entre as linhas para não dar erro de indentação
spark = SparkSession.builder \
    .appName("PostgresToIceberg") \
    .config("spark.sql.catalog.nessie", "org.apache.iceberg.spark.SparkCatalog") \
    .config("spark.sql.catalog.nessie.catalog-impl", "org.apache.iceberg.nessie.NessieCatalog") \
    .config("spark.sql.catalog.nessie.uri", "http://nessie:19120/api/v1") \
    .config("spark.sql.catalog.nessie.ref", "main") \
    .config("spark.sql.catalog.nessie.warehouse", "s3a://iceberg-bucket/") \
    .config("spark.sql.catalog.nessie.io-impl", "org.apache.iceberg.aws.s3.S3FileIO") \
    .config("spark.sql.catalog.nessie.s3.endpoint", "http://minio:9000") \
    .config("spark.sql.catalog.nessie.s3.path-style-access", "true") \
    .config("spark.sql.catalog.nessie.s3.access-key-id", "baptista") \
    .config("spark.sql.catalog.nessie.s3.secret-access-key", "cygnusX-1book2") \
    .config("spark.sql.defaultCatalog", "nessie") \
    .getOrCreate()

# Teste de leitura do Postgres
df_postgres = spark.read \
    .format("jdbc") \
    .option("url", "jdbc:postgresql://postgres-db:5432/airflow_db") \
    .option("dbtable", "dag_run") \
    .option("user", "baptista") \
    .option("password", "cygnusX-1book2") \
    .option("driver", "org.postgresql.Driver") \
    .load()

print("🎯 Se você está vendo isso, a SparkSession e a leitura do Postgres funcionaram!")
df_postgres.show(5)

# Agora sim, a gravação final no Iceberg
spark.sql("CREATE NAMESPACE IF NOT EXISTS nessie.silver_layer")
df_postgres.writeTo("nessie.silver_layer.airflow_dag_runs").createOrReplace()
print("🚀 SUCESSO TOTAL! Dados no Iceberg.")
```

```python
# 1. Criar o namespace (como se fosse um banco de dados) no Nessie
spark.sql("CREATE NAMESPACE IF NOT EXISTS nessie.silver_layer")

# 2. Gravar os dados lidos do Postgres no formato Iceberg
# Isso vai criar os metadados no Nessie e os arquivos Parquet no MinIO
df_postgres.writeTo("nessie.silver_layer.airflow_dag_runs").createOrReplace()

print("✅ Dados gravados com sucesso no Iceberg!")

# 3. Teste de Fogo: Ler de volta usando SQL (direto do Catálogo)
print("Lendo agora da tabela Iceberg:")
spark.sql("SELECT dag_id, run_type, state, start_date FROM nessie.silver_layer.airflow_dag_runs LIMIT 5").show()
```
